## 1. Training and Visualizing a Decision Tree

Decision trees require very little data preparation. They don't require feature scaling or centering.

```python
iris_data = load_iris(as_frame=True)
X, y = iris_data['data'][['petal length (cm)', 'petal width (cm)']], iris_data['target']

tree_clf = DecisionTreeClassifier(max_depth = 2, random_state = 32)
tree_clf.fit(X, y)
```

## 4. The CART Training Algorithm


<br>

The cost function J is computed many times per split, and this repeats recursively at every node.
- Within a single node — CART doesn't magically know the best split. It does a brute-force search: for every feature k and every candidate threshold t_k, it computes J(k, t_k) and picks the pair that minimizes it. So J is evaluated roughly (number of features × number of candidate thresholds) times at that one node — not just once.
- Across the tree — once the best split is found and the data is divided, each child subset becomes a new node, and the entire brute-force search runs again on it. This recurses until a stopping condition is hit (max_depth, min_samples_split, no impurity-reducing split, etc.).

## 8. Regression

Decision trees can perform regression tasks and deal with non-linear data.

```python
rng = np.random.default_rng(seed=32)
X_quad = rng.random((200, 1)) - 0.5  # a single random input feature
y_quad = X_quad ** 2 + 0.025 * rng.standard_normal((200, 1))

tree_reg = DecisionTreeRegressor(max_depth=2, random_state=32)
tree_reg.fit(X_quad, y_quad)
```
