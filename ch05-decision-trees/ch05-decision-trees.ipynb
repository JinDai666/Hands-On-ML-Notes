# Chapter 5. Decision Trees

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from graphviz import Source
```

## 1. Training and Visualizing a Decision Tree

Decision trees require very little data preparation. They don't require feature scaling or centering.

```python
iris_data = load_iris(as_frame=True)
X, y = iris_data['data'][['petal length (cm)', 'petal width (cm)']], iris_data['target']

tree_clf = DecisionTreeClassifier(max_depth = 2, random_state = 32)
tree_clf.fit(X, y)
```

```python
export_graphviz(tree_clf, 
                out_file = "/Users/jin/Desktop/Data_Scientist/Hands-On-ML-Notes/ch05-decision-trees/iris_tree.dot",
                feature_names = X.columns.values,
                class_names = iris_data.target_names,
                filled = True)
```
